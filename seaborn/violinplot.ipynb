# Violin Plots in Seaborn

A **Violin Plot** combines a box plot with a **Kernel Density Estimate (KDE)** on each side.
Unlike box plots, violin plots reveal whether the data has multiple peaks (multimodal distribution) or unusual skewness.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data_path = os.path.join("datasets", "tips.csv")
if os.path.exists(data_path):
    tips = pd.read_csv(data_path)
else:
    tips = sns.load_dataset("tips")

tips.head()
```

## 1. Standard Violin Plot with Quartiles


```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=tips, x="day", y="total_bill", inner="quart", palette="Pastel1")
plt.title("Violin Plot Showing Density and Quartiles Across Days")
plt.show()
```

## 2. Split Violin Plot for Two-Level Hue
When a hue has exactly two categories (e.g. Male / Female), `split=True` draws each group on one half of the violin.


```python
plt.figure(figsize=(9, 5))
sns.violinplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    split=True,
    inner="box",
    palette="coolwarm"
)
plt.title("Split Violin Plot: Male vs Female Bill Distribution")
plt.show()
```
