# Heatmaps in Seaborn

A **Heatmap** visualizes 2D matrices as color-encoded grids. It is the primary tool in data science for:
1. **Correlation Matrices**: Identifying strength and direction of linear relationships between numeric features.
2. **Pivot Tables**: Summarizing cross-tabulated metrics across two categorical variables.


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()

tips_path = os.path.join("datasets", "tips.csv")
if os.path.exists(tips_path):
    tips = pd.read_csv(tips_path)
else:
    tips = sns.load_dataset("tips")

flights_path = os.path.join("datasets", "flights.csv")
if os.path.exists(flights_path):
    flights = pd.read_csv(flights_path)
else:
    flights = sns.load_dataset("flights")
```

## 1. Correlation Matrix Heatmap


```python
numeric_tips = tips.select_dtypes(include="number")
corr = numeric_tips.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    corr,
    annot=True,
    fmt=".3f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    square=True
)
plt.title("Correlation Matrix Heatmap (Tips Dataset)")
plt.show()
```

## 2. Pivot Table Heatmap: Monthly Airline Passengers (1949 - 1960)


```python
flight_matrix = flights.pivot(index="month", columns="year", values="passengers")

plt.figure(figsize=(11, 6))
sns.heatmap(
    flight_matrix,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=0.4
)
plt.title("Monthly Flight Passengers (1949 - 1960)")
plt.xlabel("Year")
plt.ylabel("Month")
plt.show()
```
