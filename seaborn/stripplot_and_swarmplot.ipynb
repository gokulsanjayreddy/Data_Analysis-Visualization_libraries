# Categorical Scatter Plots: Stripplots and Swarmplots

When one of the variables is categorical, standard scatter plots cause points to overlap into uninformative vertical or horizontal lines.
Seaborn provides two specialized plots for this:
1. **Strip Plot (`sns.stripplot`)**: Adds random horizontal dispersion (`jitter`) to separate overlapping points.
2. **Swarm Plot (`sns.swarmplot`)**: Employs an algorithmic beeswarm arrangement so no points overlap, reflecting actual density.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="ticks")

data_path = os.path.join("datasets", "tips.csv")
if os.path.exists(data_path):
    tips = pd.read_csv(data_path)
else:
    tips = sns.load_dataset("tips")

tips.head()
```

## 1. Strip Plot (With and Without Jitter)
Observe how `jitter=True` prevents overlapping points from collapsing onto a single line.


```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Without jitter
sns.stripplot(data=tips, x="day", y="total_bill", jitter=False, ax=axes[0])
axes[0].set_title("Strip Plot Without Jitter (Overlapping Points)")

# With jitter
sns.stripplot(data=tips, x="day", y="total_bill", jitter=0.25, palette="Dark2", ax=axes[1])
axes[1].set_title("Strip Plot With Jitter (Reveals Point Spread)")

plt.tight_layout()
plt.show()
```

## 2. Strip Plot with Hue and Dodge
Using `dodge=True` places groups side-by-side instead of intermingling them.


```python
plt.figure(figsize=(8, 5))
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    dodge=True,
    jitter=0.2,
    palette="Accent"
)
plt.title("Bill Distribution per Day Split by Gender (Dodge=True)")
plt.show()
```

## 3. Swarm Plot (Beeswarm Algorithm)
Adjusts points along the categorical axis so that no points overlap.


```python
plt.figure(figsize=(9, 5))
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    dodge=True,
    palette="tab10",
    size=5
)
plt.title("Swarm Plot of Total Bill Across Days by Sex")
plt.show()
```
