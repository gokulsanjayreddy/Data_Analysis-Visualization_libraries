# Pair Plots in Seaborn

A **Pair Plot** (`sns.pairplot`) constructs a matrix of subplots visualizing pairwise relationships between all numeric variables in a DataFrame.
- **Diagonal plots**: Univariate distribution of each feature (histogram or KDE).
- **Off-diagonal plots**: Bivariate scatter plots showing correlations.
- **`corner=True`**: Omits redundant symmetric comparisons above the diagonal.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="ticks")

iris = sns.load_dataset("iris")
iris.head()
```

## 1. Comprehensive Pair Plot on Iris Dataset


```python
sns.pairplot(
    iris,
    hue="species",
    palette="Set2",
    diag_kind="kde",
    markers=["o", "s", "D"]
)
plt.show()
```

## 2. Corner Pair Plot on Tips Dataset
`corner=True` displays only the lower triangular portion of the grid.


```python
data_path = os.path.join("datasets", "tips.csv")
if os.path.exists(data_path):
    tips = pd.read_csv(data_path)
else:
    tips = sns.load_dataset("tips")

sns.pairplot(
    tips,
    hue="time",
    corner=True,
    palette="coolwarm"
)
plt.show()
```
