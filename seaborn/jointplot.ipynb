# Joint Plots in Seaborn

A **Joint Plot** (`sns.jointplot`) combines a bivariate relationship plot in the center with univariate marginal distribution plots (histograms/KDE) along the top and right axes.
Available kinds include:
- `'scatter'`: Standard scatter plot with marginal histograms.
- `'hex'`: Hexagonal binning for dense, overlapping data points.
- `'kde'`: 2D bivariate contour density plot.
- `'reg'`: Scatter plot with linear regression trendline and marginal KDEs.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="white")

tips_path = os.path.join("datasets", "tips.csv")
if os.path.exists(tips_path):
    tips = pd.read_csv(tips_path)
else:
    tips = sns.load_dataset("tips")

tips.head()
```

## 1. Scatter Joint Plot with Hue


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    palette="Set1",
    height=6
)
plt.show()
```

## 2. Hexagonal Binning (`kind='hex'`)


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="hex",
    color="#4CB391",
    height=6
)
plt.show()
```

## 3. Regression Joint Plot (`kind='reg'`)


```python
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip",
    kind="reg",
    color="purple",
    height=6
)
plt.show()
```
