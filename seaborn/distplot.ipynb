# Distribution Plots in Seaborn

Analyzing the distribution of numerical variables is fundamental to data analysis.
Seaborn provides:
- `sns.histplot()`: Modern, versatile histogram with bin count, KDE curve, and stacking options.
- `sns.kdeplot()`: Continuous Kernel Density Estimate curve representing probability density.
- `sns.rugplot()`: Small tick marks along an axis showing every individual observation.
- `sns.distplot()`: Legacy function combining histogram and KDE (deprecated in Seaborn 0.11+).


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

tips_path = os.path.join("datasets", "tips.csv")
if os.path.exists(tips_path):
    tips = pd.read_csv(tips_path)
else:
    tips = sns.load_dataset("tips")

tips.head()
```

## 1. Histogram with KDE Curve (`sns.histplot`)


```python
plt.figure(figsize=(8, 5))
sns.histplot(
    data=tips,
    x="total_bill",
    kde=True,
    bins=25,
    color="royalblue"
)
plt.title("Histogram and KDE of Total Bill")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")
plt.show()
```

## 2. Multi-group Distribution with Multiple='stack'


```python
plt.figure(figsize=(8, 5))
sns.histplot(
    data=tips,
    x="total_bill",
    hue="sex",
    multiple="stack",
    palette="pastel"
)
plt.title("Stacked Total Bill Distribution by Gender")
plt.show()
```

## 3. Kernel Density Estimation with Fill (`sns.kdeplot`)


```python
plt.figure(figsize=(8, 5))
sns.kdeplot(
    data=tips,
    x="total_bill",
    hue="time",
    fill=True,
    common_norm=False,
    palette="magma"
)
plt.title("Kernel Density Estimate: Lunch vs Dinner")
plt.show()
```

## 4. Combined Histogram, KDE, and Rug Plot


```python
plt.figure(figsize=(8, 5))
sns.histplot(data=tips, x="tip", kde=True, color="darkgreen", stat="density")
sns.rugplot(data=tips, x="tip", color="red", height=0.08)
plt.title("Tip Distribution with Density Curve and Rug Ticks")
plt.show()
```
