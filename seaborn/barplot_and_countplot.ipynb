# Bar Plots and Count Plots in Seaborn

- **Bar Plot (`sns.barplot`)**: Displays statistical estimates (default: mean) of a continuous variable for each category, with error bars showing uncertainty (confidence interval).
- **Count Plot (`sns.countplot`)**: Displays the frequency (count) of occurrences for each categorical value (similar to a histogram for categorical data).


```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

titanic_path = os.path.join("datasets", "titanic.csv")
if os.path.exists(titanic_path):
    titanic = pd.read_csv(titanic_path)
else:
    titanic = sns.load_dataset("titanic")

titanic.head()
```

## 1. Bar Plot with Statistical Estimators and Error Bars


```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    estimator=np.mean,
    errorbar="ci",
    palette="Blues"
)
plt.title("Average Fare by Passenger Class and Sex (with 95% CI)")
plt.ylabel("Average Ticket Fare ($)")
plt.show()
```

## 2. Count Plot: Categorical Frequency Distribution


```python
plt.figure(figsize=(8, 5))
sns.countplot(
    data=titanic,
    x="class",
    hue="survived",
    palette="Set2"
)
plt.title("Passenger Survival Count by Class (0 = Died, 1 = Survived)")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")
plt.show()
```
