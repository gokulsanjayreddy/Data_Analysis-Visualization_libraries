# Scatter Plots in Seaborn

A **Scatter Plot** visualizes the relationship and correlation between two continuous numerical variables on a 2D Cartesian plane.
Seaborn's `sns.scatterplot()` enhances standard scatter plots by easily mapping additional dimensions to semantic properties such as:
- `hue`: Color-coding by category or continuous scale.
- `style`: Marker shape differentiation across categorical subsets.
- `size`: Scaling marker sizes based on a numerical or ordinal variable.
- `palette`: Pre-defined or customized color themes.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual theme
sns.set_theme(style="whitegrid")

# Load Tips dataset
data_path = os.path.join("datasets", "tips.csv")
if os.path.exists(data_path):
    tips = pd.read_csv(data_path)
else:
    tips = sns.load_dataset("tips")

print("Tips dataset shape:", tips.shape)
tips.head()
```

## 1. Basic Scatter Plot
Examines whether higher bills correspond to larger tips.


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

## 2. Adding Categorical Dimension using Hue and Style
Distinguish customers by `sex` (color) and `smoker` status (marker style).


```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    style="smoker",
    palette="Set1",
    s=70,
    alpha=0.85
)
plt.title("Bill vs Tip Segmented by Gender and Smoker Status")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
plt.show()
```

## 3. Multi-dimensional Scatter Plot with Size
Incorporate party size (`size`) to scale the diameter of each point.


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="day",
    size="size",
    sizes=(25, 250),
    palette="viridis",
    alpha=0.8
)
plt.title("Total Bill vs Tip by Day and Party Size")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")
sns.despine()
plt.show()
```
