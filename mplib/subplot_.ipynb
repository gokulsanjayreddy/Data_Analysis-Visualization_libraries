## Subplot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.subplot(1, 2, 1) # 1 row, 2 columns, 1st subplot -> has 1 row and 2 cols and the subplot no is 1 
plt.plot(x, y)
plt.title('Line chart')

plt.subplot(1, 2, 2) # 1 row, 2 columns, 2nd subplot -> here the subplot no is 2
plt.bar(x, y)
plt.title('Bar Chart')

plt.tight_layout()#used to adjust the space betweeen layout
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]
# fig, ax = plt.subplots(nrows, ncols , figsize = (width,height))
fig, ax = plt.subplots(1, 2, figsize = (10, 5))

x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

ax[0].plot(x, y)
ax[0].set_title('Line Plot')

ax[1].bar(x, y,color = 'green')#we can also pass color here
ax[1].set_title('Bar Chart')
fig.suptitle('Comparison between line plot and bar chart') #S set the title of the total figure or the total canvas
plt.tight_layout()
# always pass the tight layout just before the show function 
plt.show()
```
