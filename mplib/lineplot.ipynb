# **Line plot**

```python
import matplotlib.pyplot as plt
x = [1,2,3,4]
y = [10,20,15,25]
plt.plot(x,y)#plots the graph of line joining the points x,y
plt.show()#we have to call this to retrive the output
```

### Understanding terms with plot

```python
import matplotlib.pyplot as plt
x = ['mon','tues','wed','thurs','fri']
y = [10,20,15,40,16]
plt.plot(x,y)
plt.xlabel('Days of week')
plt.ylabel('sales per day')
plt.title('BAKERY SALES')
plt.show()
```
