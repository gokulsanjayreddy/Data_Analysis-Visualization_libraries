## **Using functions of pyplot**

```python
# plt.plot(x,y,color = 'color name',linestyle = 'linestyle',linewidth = value , marker = 'marker symbol' , label = 'label name')
# plt.xlabel('xyz') for labelling x axis and similarly ylabel
# plt.title('xyz') for setting the title of the graph
# plt.legend(location , fontsize)-upperleft or lowerright
# plt.grid( color,linestyle,linewidth) -> for the grid behind the graph
# plt.xlim(start,end) -> to control the limits of  x axis and similarly ylim for yaxis 
# plt.xticks(old,new) and yticks to represent or change the ticks of X and Y axis respectively
import matplotlib.pyplot as plt
x = ['mon','tues','wed','thurs','fri']
y = [10,5,60,40,2]
plt.plot(x,y,color = 'blue',linestyle = '--',linewidth = 2 , marker = 'o' , label = 'Sales data')
plt.xlabel('days of week')
plt.ylabel('sales')
plt.title('Shop sales in the week')
plt.grid(color ='grey',linestyle = ':',linewidth = 1)
plt.legend(loc ='upper left' ,fontsize= 12)
plt.xticks(['mon','tues','wed','thurs','fri'],['m','t','w','th','f'])
plt.show()
```
